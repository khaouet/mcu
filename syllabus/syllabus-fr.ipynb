### Activité 2 — Architecture ARM Cortex-M3 (30 min)

#### 2.1 Vue d'ensemble

Le **Cortex-M3** est un cœur 32 bits RISC développé par ARM, optimisé pour les systèmes embarqués.

| Élément | Description |
|---|---|
| Architecture | ARMv7-M (Harvard modifiée) |
| Largeur | 32 bits |
| Pipeline | 3 étages (Fetch – Decode – Execute) |
| Jeu d'instructions | Thumb-2 (16 et 32 bits) |
| Registres généraux | R0 à R12 |
| Registres spéciaux | SP, LR, PC, PSR |
| Interruptions | NVIC intégré (jusqu'à 240 IRQ) |
| Horloge max (STM32F103) | 72 MHz |
| MPU | Optionnelle |
| Debug | SWD / JTAG |

#### 2.2 Schéma fonctionnel du cœur

```
┌─────────────────────────────────────────────┐
│              CORTEX-M3 CORE                 │
│  ┌───────┐   ┌────────┐    ┌──────────────┐ │
│  │ Fetch │──▶│ Decode │──▶│   Execute    │ │
│  └───────┘   └────────┘    └──────────────┘ │
│        │                        │           │
│        ▼                        ▼           │
│   ┌─────────┐             ┌──────────┐      │
│   │  Flash  │             │   SRAM   │      │
│   └─────────┘             └──────────┘      │
│        │                        │           │
│        └───────┬────────────────┘           │
│                ▼                            │
│         ┌──────────────┐                    │
│         │  Bus Matrix  │                    │
│         └──────┬───────┘                    │
│                │                            │
│     ┌──────────┼─────────────┐              │
│     ▼          ▼             ▼              │
│  ┌──────┐  ┌───────┐    ┌─────────┐         │
│  │ NVIC │  │ Debug │    │  Periph │         │
│  └──────┘  └───────┘    └─────────┘         │
└─────────────────────────────────────────────┘
```

#### 2.3 Modes et privilèges

| Mode | Description |
|---|---|
| **Thread** | Exécution normale |
| **Handler** | Exécution d'une ISR |

| Niveau | Description |
|---|---|
| **Privilégié** | Accès complet |
| **Non privilégié** | Accès limité (OS) |

**États d'énergie :** Running – Sleep – Deep-Sleep

## 🎓 Cours (1h30)

### Activité 1 — Vue d'ensemble du STM32F103C6T6 (30 min)

| Caractéristique | Valeur |
|---|---|
| Cœur | ARM Cortex-M3 @ 72 MHz |
| Flash | 32 Ko |
| SRAM | 10 Ko |
| GPIO | 37 (LQFP48) |
| Timers | 3 (TIM1, TIM2, TIM3) |
| ADC | 2 × 12 bits, 10 canaux |
| UART | 2 (USART1, USART2) |
| SPI | 2 (SPI1, SPI2) |
| I2C | 1 (I2C1) |
| USB | 1 (Full-Speed) |
| CAN | 1 |
| Alimentation | 2.0 V – 3.6 V |
| Boîtier | LQFP48 |

### Activité 2 — Bus et mappe mémoire (30 min)

**Bus principaux :**
- **AHB** (Advanced High-performance Bus) — 72 MHz : Flash, SRAM, DMA, GPIO
- **APB1** — 36 MHz : TIM2/3, USART2, I2C1, SPI2
- **APB2** — 72 MHz : TIM1, USART1, SPI1, ADC, GPIO, EXTI

**Mappe mémoire (extrait) :**

| Adresse | Contenu |
|---|---|
| 0x0000_0000 | Code (Flash alias) |
| 0x0800_0000 | Flash (32 Ko) |
| 0x2000_0000 | SRAM (10 Ko) |
| 0x4000_0000 | Périphériques APB1/APB2/AHB |
| 0xE000_0000 | NVIC, SysTick (privé) |

### Activité 3 — Horloges (30 min)

**Sources d'horloge :**
- **HSI** : 8 MHz interne (±1 %)
- **HSE** : 4–16 MHz externe (souvent 8 MHz sur Blue Pill)
- **PLL** : multiplie HSI ou HSE pour atteindre 72 MHz
- **LSI/LSE** : 40 kHz / 32.768 kHz pour RTC et IWDG

**Arbre d'horloge simplifié :**
```
HSE(8MHz) ──▶ PLL ×9 ──▶ SYSCLK(72MHz)
                            │
                            ├── AHB presc=1 ──▶ HCLK(72MHz)
                            ├── APB1 presc=2 ──▶ PCLK1(36MHz)
                            └── APB2 presc=1 ──▶ PCLK2(72MHz)
```

````python
// Extrait SystemClock_Config généré par CubeMX
void SystemClock_Config(void)
{
  RCC_OscInitTypeDef RCC_OscInitStruct = {0};
  RCC_ClkInitTypeDef RCC_ClkInitStruct = {0};

  RCC_OscInitStruct.OscillatorType = RCC_OSCILLATORTYPE_HSE;
  RCC_OscInitStruct.HSEState = RCC_HSE_ON;
  RCC_OscInitStruct.HSEPredivValue = RCC_HSE_PREDIV_DIV1;
  RCC_OscInitStruct.HSIState = RCC_HSI_ON;
  RCC_OscInitStruct.PLL.PLLState = RCC_PLL_ON;
  RCC_OscInitStruct.PLL.PLLSource = RCC_PLLSOURCE_HSE;
  RCC_OscInitStruct.PLL.PLLMUL = RCC_PLL_MUL9; // 8 MHz × 9 = 72 MHz
  HAL_RCC_OscConfig(&RCC_OscInitStruct);

  RCC_ClkInitStruct.ClockType = RCC_CLOCKTYPE_HCLK | RCC_CLOCKTYPE_SYSCLK
                              | RCC_CLOCKTYPE_PCLK1 | RCC_CLOCKTYPE_PCLK2;
  RCC_ClkInitStruct.SYSCLKSource = RCC_SYSCLKSOURCE_PLLCLK;
  RCC_ClkInitStruct.AHBCLKDivider = RCC_SYSCLK_DIV1;   // 72 MHz
  RCC_ClkInitStruct.APB1CLKDivider = RCC_HCLK_DIV2;    // 36 MHz
  RCC_ClkInitStruct.APB2CLKDivider = RCC_HCLK_DIV1;    // 72 MHz
  HAL_RCC_ClockConfig(&RCC_ClkInitStruct, FLASH_LATENCY_2);
}
````

# 📘 Semaine 4 — EXTI + NVIC

**Durée :** 4h30

## 🎯 Objectifs
- Comprendre le mécanisme d'interruption
- Configurer EXTI sur un GPIO

## 🎓 Cours (1h30)

### Activité 1 — Interruptions (45 min)

- **IRQ** : Interrupt Request (matériel)
- **ISR** : Interrupt Service Routine
- **NVIC** : Nested Vectored Interrupt Controller
- **Priorités** : 16 niveaux (4 bits), préemption + sous-priorité
- **Table des vecteurs** en Flash (début de la mémoire)

### Activité 2 — EXTI (45 min)

- 16 lignes EXTI (EXTI0–EXTI15)
- Chaque ligne mappée sur un GPIO via AFIO_EXTICR
- Fronts : montant, descendant, les deux
- Registres : IMR, EMR, RTSR, FTSR, PR (pending)

**Exemple HAL :**
```c
// Dans main.c
HAL_NVIC_SetPriority(EXTI0_IRQn, 1, 0);
HAL_NVIC_EnableIRQ(EXTI0_IRQn);

// Callback
void HAL_GPIO_EXTI_Callback(uint16_t GPIO_Pin)
{
  if (GPIO_Pin == GPIO_PIN_0)
    HAL_GPIO_TogglePin(GPIOC, GPIO_PIN_13);
}

## 🎓 Cours (1h30)

### Activité 1 — Principe PWM (45 min)

**Formules :**
```
F_pwm = F_clk / ((PSC + 1) × (ARR + 1))
Duty (%) = CCR / (ARR + 1) × 100
```

**Modes PWM :**
- Mode 1 : actif tant que CNT < CCR
- Mode 2 : inverse
- Alignement : edge ou center

### Activité 2 — Applications (45 min)

| Application | Fréquence | Plage duty |
|---|---|---|
| LED (dimming) | 1–10 kHz | 0–100 % |
| Servo moteur | 50 Hz | 5–10 % (1–2 ms) |
| Moteur DC | 10–20 kHz | 0–100 % |
| Buzzer | 2–4 kHz | 50 % |

```c
// Régler duty cycle
__HAL_TIM_SET_COMPARE(&htim2, TIM_CHANNEL_1, 500);

# 📘 Semaine 8 — ADC (principes)

**Durée :** 4h30

## 🎯 Objectifs
- Comprendre le principe de la conversion analogique-numérique
- Configurer l'ADC 12 bits du STM32
- Lire un potentiomètre / capteur analogique

## 🎓 Cours (1h30)

### Activité 1 — Principes ADC (45 min)

**Formules :**
```
Résolution = V_ref / 2^n
Code = (V_in / V_ref) × 4095  (12 bits)
V_in = Code × V_ref / 4095
```

**Caractéristiques ADC STM32F103 :**

| Paramètre | Valeur |
|---|---|
| Résolution | 12 bits (4096 niveaux) |
| Canaux | 10 externes + 2 internes |
| V_ref | VDDA (2.4–3.6 V) |
| Temps de conversion | 1 µs @ 14 MHz |
| Modes | Simple, scan, continu, discontinu |
| Déclencheurs | Software, timer, EXTI |

### Activité 2 — Configuration HAL (45 min)

```c
HAL_ADC_Start(&hadc1);
HAL_ADC_PollForConversion(&hadc1, 100);
uint16_t val = HAL_ADC_GetValue(&hadc1);
float tension = val * 3.3f / 4095.0f;
HAL_ADC_Stop(&hadc1);
```
