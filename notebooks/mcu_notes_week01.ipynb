## 🔹 Activité 2 — Notions fondamentales (15 min)

### 📖 Cours — Microprocesseur vs Microcontrôleur

| Critère | Microprocesseur (µP) | Microcontrôleur (µC) |
|---|---|---|
| **Définition** | CPU seul | CPU + mémoires + périphériques |
| **Mémoire** | Externe (RAM, ROM) | Interne (Flash, SRAM) |
| **Périphériques** | Externes (chipset) | Intégrés (GPIO, Timers, ADC, UART…) |
| **Coût** | Élevé | Faible |
| **Consommation** | Élevée | Faible |
| **Usage typique** | PC, serveurs | Embarqué, IoT, automobile |
| **Exemples** | Intel Core i7, AMD Ryzen | STM32, PIC, ATmega, ESP32 |

### 🧠 À retenir

> Un **microcontrôleur** est un **système complet sur puce** (*System on Chip* — SoC) :  
> **CPU + Flash + SRAM + Périphériques + Bus internes**.

Un serveur ou un ordinateur portable utilise un microprocesseur ; un STM32 utilise un microcontrôleur.

### ⚠️ Piège fréquent

- **microcontrôleur** : CPU + mémoires + périphériques intégrés

Ne confondez pas :- **microprocesseur** : CPU seule, avec mémoires externes

## 🔹 Activité 3 — Architecture ARM Cortex-M3 (30 min)

### 📖 3.1 — Vue d'ensemble

Le **Cortex-M3** est un cœur **32 bits RISC** conçu par **ARM Holdings** et **licencié** aux fabricants (ST, NXP, TI, Nordic…).

> Référence RM0008 : architecture du cœur Cortex-M3, NVIC, modes de fonctionnement et organisation des périphériques STM32.

| Élément | Description |
|---|---|
| **Architecture** | ARMv7-M (Harvard modifiée) |
| **Largeur** | 32 bits |
| **Pipeline** | 3 étages : Fetch → Decode → Execute |
| **Jeu d'instructions** | Thumb-2 (mix 16 / 32 bits) |
| **Registres généraux** | R0 à R12 |
| **Registres spéciaux** | SP (R13), LR (R14), PC (R15), PSR |
| **Interruptions** | NVIC intégré (jusqu'à 240 IRQ, 16 priorités) |
| **Horloge max** | 72 MHz (sur STM32F103) |
| **Debug** | SWD / JTAG |
| **FPU** | Optionnelle |

> 🔎 **RISC** = *Reduced Instruction Set Computer* : instructions simples, exécution rapide, pipeline efficace.

### 📖 3.2 — Pipeline 3 étages

```
Cycle :      1     2     3     4     5     6
           ┌─────┬─────┬─────┐
Fetch      │ I1  │ I2  │ I3  │ I4  │ I5  │ I6  │
           └─────┴─────┴─────┴─────┴─────┴─────┘
                 ┌─────┬─────┬─────┐
Decode           │ I1  │ I2  │ I3  │ I4  │ I5  │
                 └─────┴─────┴─────┴─────┴─────┘
                       ┌─────┬─────┬─────┐
Execute                │ I1  │ I2  │ I3  │ I4  │
                       └─────┴─────┴─────┴─────┘
```

**Idée clé :** pendant qu'une instruction s'exécute, la suivante se décode, et la suivante encore se charge. → **1 instruction par cycle** en régime permanent.

**Conséquence :** un branchement (saut) vide le pipeline → *branch penalty*.

### 📖 3.3 — Registres du Cortex-M3

| Registre | Rôle |
|---|---|
| **R0 – R12** | Registres généraux (données) |
| **R13 (SP)** | Stack Pointer — pointeur de pile |
| **R14 (LR)** | Link Register — adresse de retour d'appel |
| **R15 (PC)** | Program Counter — adresse de l'instruction courante |
| **PSR** | Program Status Register (flags N, Z, C, V, mode, IRQ) |
| **PRIMASK** | Masque global des interruptions |
| **FAULTMASK** | Masque toutes les exceptions sauf NMI |
| **BASEPRI** | Masque par niveau de priorité |
| **CONTROL** | Mode (Thread/Handler), privilège, choix de pile |

### 📖 3.4 — Modes et privilèges

**Deux modes d'exécution :**

| Mode | Description |
|---|---|
| **Thread** | Exécution normale (programme principal) |
| **Handler** | Exécution d'une routine d'interruption (ISR) |

**Deux niveaux de privilège :**

| Niveau | Description |
|---|---|
| **Privilégié** | Accès complet aux ressources |
| **Non privilégié** | Accès restreint (utile pour un OS) |

**États d'énergie :**

| État | Description |
|---|---|
| **Running** | CPU actif |
| **Sleep** | Horloge CPU arrêtée, périphériques actifs |
| **Deep-Sleep** | Horloges principales arrêtées, réveil par IRQ |

### 📖 3.5 — Le NVIC (Nested Vectored Interrupt Controller)

Le **NVIC** est intégré au cœur Cortex-M3. Il gère :
- **240 IRQ** maximum (le STM32F103C6T6 en expose 43)
- **16 niveaux de priorité** (codés sur 4 bits)
- **Préemption** (une IRQ prioritaire peut interrompre une ISR en cours)
- **Vecteurs** : chaque IRQ a son adresse dans la *vector table*

**Registres NVIC principaux :**

| Registre | Rôle |
|---|---|
| `ISER` | Interrupt Set-Enable |
| `ICER` | Interrupt Clear-Enable |
| `ISPR` | Interrupt Set-Pending |
| `ICPR` | Interrupt Clear-Pending |
| `IPR` | Interrupt Priority |

**Schéma simplifié :**
```
Périphérique ──▶ NVIC ──▶ Cœur Cortex-M3
 (EXTI, TIM, ADC…)   │
                     ├─ priorité
                     ├─ préemption
                     └─ vecteur
```

- Le **NVIC** gère les interruptions et les priorités.

### ✅ À retenir- Les registres **R13, R14, R15** sont essentiels pour la pile, le retour d'appel et le compteur d'instructions.

- Le **Cortex-M3** est un cœur **32 bits RISC**.- Le **pipeline** est en **3 étages** : Fetch → Decode → Execute.

### 💻 Simulation Python — Pipeline simplifié (10 min)

Le code ci-dessous **simule un pipeline 3 étages** pour visualiser l'exécution de 6 instructions.

👉 **Exécute la cellule** avec `Shift + Entrée`.

````python
# Simulation d'un pipeline 3 étages
instructions = [f"I{i+1}" for i in range(8)]
stages = ["Fetch", "Decode", "Execute"]

print("Cycle :", "\t".join(f"{i+1}" for i in range(len(instructions) + 2)))
print("-" * 70)

for s, stage in enumerate(stages):
    row = ["  .  "] * (len(instructions) + 2)
    for i, instr in enumerate(instructions):
        row[i + s] = f" {instr:<3} "
    print(f"{stage:<8}", "\t".join(row))

print("\n📌 Observation :") 
print("  → 1 instruction terminée par cycle en régime permanent.")
print("  → Un branchement provoquerait un 'branch penalty' (vidage du pipeline).")
````

### 💻 Vérification rapide (Linux / macOS)

Exécute la cellule suivante pour vérifier si un ST-Link est détecté sur ton port USB.

````python
# Vérifie la présence d'un ST-Link branché en USB
import platform, subprocess

system = platform.system()
print(f"🖥️  Système détecté : {system}\n")

try:
    if system == "Linux":
        out = subprocess.run(["lsusb"], capture_output=True, text=True).stdout
    elif system == "Darwin":
        out = subprocess.run(["system_profiler", "SPUSBDataType"], capture_output=True, text=True).stdout
    elif system == "Windows":
        out = subprocess.run(
            [
                "powershell",
                "-NoProfile",
                "-Command",
                "Get-PnpDevice -PresentOnly | Where-Object { $_.FriendlyName -match 'ST-Link|STLink' } | Select-Object -ExpandProperty FriendlyName",
            ],
            capture_output=True,
            text=True,
        ).stdout
    else:
        out = ""

    if "ST-Link" in out or "STLink" in out or "0483" in out:
        print("✅ ST-Link détecté !")
    else:
        print("⚠️  Aucun ST-Link détecté.")
        print("   → Vérifie le câble USB, les drivers et la connexion SWD.")

except Exception as e:
    print("ℹ️  Impossible d'exécuter la détection :", e)
    print("   → Ce notebook peut tourner hors ligne ou sur Colab.")
````

## 🔹 Activité 6 — Premier projet "Blink LED" (démo guidée — 15 min)

### 🎯 Objectif
Faire clignoter la LED intégrée (PC13) à 1 Hz.

### 📋 Étapes dans STM32CubeIDE

1. **File → New → STM32 Project**
2. **MCU Selector** : taper `STM32F103C6Tx`
3. **Nom du projet** : `Blink_LED`
4. **Pinout & Configuration** :
   - Cliquer sur `PC13` → `GPIO_Output`
5. **Clock Configuration** : HSI 8 MHz par défaut (suffisant)
6. **Project Manager → Generate Code**
7. Ouvrir `Core/Src/main.c` et ajouter la boucle

````python
// Extrait C — à insérer dans Core/Src/main.c
// Section : USER CODE BEGIN WHILE

while (1)
{
    HAL_GPIO_TogglePin(GPIOC, GPIO_PIN_13);  // Bascule PC13
    HAL_Delay(500);                          // Attente 500 ms
    /* USER CODE END WHILE */
}
````

### 💡 Explication du code

| Ligne | Rôle |
|---|---|
| `HAL_GPIO_TogglePin(GPIOC, GPIO_PIN_13)` | Inverse l'état du pin PC13 |
| `HAL_Delay(500)` | Attend 500 ms (bloquant) |

**Calcul de la fréquence :**  
```
T = 500 ms (ON) + 500 ms (OFF) = 1000 ms
F = 1 / T = 1 Hz
```

> ⚠️ **À retenir :** la LED du Blue Pill est connectée sur **PC13** et est souvent active à l'état bas. Il faut donc vérifier la polarité logique du montage si le comportement semble inversé.

- La LED peut sembler inversée selon la version de la carte, la polarité du montage et la configuration GPIO.

### ✅ Point clé du TP- `HAL_Delay(500)` donne une période de 1 s complète pour un clignotement à 1 Hz.
- `HAL_GPIO_TogglePin()` inverse l'état du GPIO.

````python
# Calcul de fréquence à partir d'un délai
def frequence_clignotement(delai_ms):
    """Retourne la fréquence de clignotement en Hz."""
    periode_s = 2 * delai_ms / 1000.0  # ON + OFF
    return 1 / periode_s

# Tests
for d in [500, 200, 100, 1000]:
    f = frequence_clignotement(d)
    print(f"Délai = {d:>4} ms  →  Période = {2*d:>4} ms  →  Fréquence = {f:.2f} Hz")
````

### 🧠 Glossaire rapide
- **µP** : microprocesseur ; CPU seule, avec mémoire externe.
- **µC** : microcontrôleur ; CPU + mémoire + périphériques intégrés.
- **GPIO** : General Purpose Input/Output ; broche configurable en entrée/sortie.
- **NVIC** : Nested Vectored Interrupt Controller ; gestion des interruptions.
- **ISR** : Interrupt Service Routine ; routine d’interruption.
- **RST** : reset ; remise à zéro du système.
- **SWD** : Serial Wire Debug ; interface de debug.
