## 📇 Fiches de révision express

### 🗂️ Fiche 1 — Formules essentielles

```
Timer :    F_timer = F_clk / ((PSC+1) × (ARR+1))
           T_timer = 1 / F_timer

PWM :      F_pwm = F_clk / ((PSC+1) × (ARR+1))
           Duty  = CCR / (ARR+1)
           CCR   = Duty × (ARR+1)
           V_moy = Duty × V_high

ADC :      Code  = (V_in / V_ref) × (2^n − 1)
           V_in  = Code × V_ref / (2^n − 1)
           Résolution = V_ref / 2^n

Horloge :  SYSCLK = HSE × PLL
           HCLK   = SYSCLK / AHB_prescaler
           PCLK1  = HCLK / APB1_prescaler
           PCLK2  = HCLK / APB2_prescaler
```

## 🔹 Aperçu du convertisseur ADC du STM32F103

### 📖 Caractéristiques principales

| Paramètre | Valeur |
|---|---|
| Résolution | 12 bits (4096 niveaux) |
| Canaux | 10 externes + 2 internes |
| V_ref | VDDA (2.4 – 3.6 V) |
| Temps de conversion | 1 µs @ 14 MHz ADC |
| Modes | Simple, scan, continu, discontinu |
| Déclencheurs | Logiciel, timer, EXTI |
| Alignement | Droite ou gauche |
| Nombre d'ADC | 2 (ADC1, ADC2) |

### 📖 Formules fondamentales

```
Résolution (LSB) = V_ref / 2^12 = V_ref / 4096

Code (0..4095)   = (V_in / V_ref) × 4095

V_in             = Code × V_ref / 4095
```

### 📖 Exemples

| V_in | Code (V_ref = 3.3 V) |
|---|---|
| 0 V | 0 |
| 0.806 mV (1 LSB) | 1 |
| 0.825 V (V_ref/4) | 1024 |
| 1.65 V (V_ref/2) | 2048 |
| 2.475 V (3/4 V_ref) | 3072 |
| 3.3 V | 4095 |

### 🐍 Simulation Python — Conversion ADC (10 min)

Manipulons les formules de conversion ADC avant la semaine 8.

````python
# ============================================================
# Simulateur ADC — Conversion code ↔ tension
# ============================================================

def tension_vers_code(v_in, v_ref=3.3, bits=12):
    """Retourne le code ADC (0..2^bits-1) pour une tension donnée."""
    code_max = 2 ** bits - 1
    code = round(v_in / v_ref * code_max)
    return max(0, min(code_max, code))

def code_vers_tension(code, v_ref=3.3, bits=12):
    """Retourne la tension correspondant à un code ADC."""
    code_max = 2 ** bits - 1
    return code * v_ref / code_max

def resolution_mv(v_ref=3.3, bits=12):
    return v_ref / (2 ** bits) * 1000

print("📊 ADC 12 bits — V_ref = 3.3 V")
print(f"  Résolution : {resolution_mv(3.3, 12):.3f} mV/LSB")
print()
print(f"{'Tension (V)':<15}{'Code ADC':<12}{'Tension reconstruite'}")
print("-" * 50)
for v in [0, 0.1, 0.5, 1.0, 1.65, 2.0, 2.5, 3.0, 3.3]:
    code = tension_vers_code(v, 3.3, 12)
    v_rec = code_vers_tension(code, 3.3, 12)
    err = (v_rec - v) * 1000
    print(f"{v:<15}{code:<12}{v_rec:.4f} V   (erreur {err:+.3f} mV)")
````
